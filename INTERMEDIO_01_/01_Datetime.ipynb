```python
from datetime import datetime  # importas datetime de la subcarpeta y de ahi te traes la "fábrica", que es datetime
now = datetime.now() # el nombre de la variable puede ser el que quieras, pero luego cuando pongas datetime. tienes que poner oblgatoriamente now
print(now.year)
print(now.month)
print(now.day)
print(now.hour)
print(now.minute)
print(now.second)

timestamp = now.timestamp() # timestamp: Es la representación de este momento concreto
print(timestamp)
```

```python
from datetime import datetime

def mostrar_reloj(tiempo_estacion):
  print(tiempo_estacion.hour)
  print(tiempo_estacion.minute)

reloj_actual = datetime.now()
timestamp = reloj_actual.timestamp() # el timestamp siempre tiene que ir acompañado de una variable y . nunca puede ir solo
print(timestamp)
mostrar_reloj(reloj_actual)
```

timedelta sirve para sumar u restar tiempo (dias,minutos,horas o semanas) a una fecha


```python
from datetime import datetime, timedelta

fecha_salida = datetime(2026,10,16)

plazo = timedelta(days=12)

logistica = timedelta(days=2)

entrega = fecha_salida + plazo
print(entrega)

limite_almacen = entrega - logistica
print(limite_almacen)
print(f"El paquete llegará correctamente al cliente en la fecha {limite_almacen}")
```

```python
from datetime import datetime, timedelta

compra_dispositivo = datetime(2025,5,20)

plazo_soporte = timedelta(days=90)

departamento_calidad = timedelta(days=5)

expiracion_soporte = compra_dispositivo + plazo_soporte
print(f"El día que expira el plazo de soporte estándar desde la fecha de compra es el día {expiracion_soporte}")

fecha_fabricacion = compra_dispositivo - departamento_calidad
print(f"La fecha de fabricación de este objeto fue a fecha {fecha_fabricacion}")
```

```python
from datetime import datetime,timedelta
hora_actual = datetime.now()
print(hora_actual.minute)

zona_azul = timedelta(minutes=120)

revisor = timedelta(minutes=15)

expiracion_ticket = hora_actual + zona_azul
print(f"La hora a la que experirá el ticket será a las {expiracion_ticket}")

ultima_vez_revisor = hora_actual - revisor
print(f"La última vez que pasó el revisor fue a las {ultima_vez_revisor}")

```
