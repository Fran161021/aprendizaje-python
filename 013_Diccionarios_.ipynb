Diccionarios. Es una colección modificable no ordenada de pares únicos de claves y valor

```python
capitales = {
'EE.UU':'Washinton D.C',
'Argentina':'Buenos Aires',
'España':'Madrid',
'Brasil':'Brasilia',
}
print(capitales['España'])
```

```python
capitales = {
'EE.UU':'Washinton D.C',
'Argentina':'Buenos Aires',
'España':'Madrid',
'Brasil':'Brasilia',
}
print(capitales.get('Alemania')) # get: cambiamos los corchetes por los paréntesis, y get se usa para verificar si una clave está en el diccionario
```

```python
capitales = {
'EE.UU':'Washinton D.C',
'Argentina':'Buenos Aires',
'España':'Madrid',
'Brasil':'Brasilia',
}
print(capitales.get('Argentina')) # En este caso esta clave si estña en el diccionario
```

```python
#print(capitales.get('Argentina'))
print(capitales.keys()) # keys: Nos muestra las llaves de ese diccionario, que son cada uno de los paises
```

```python
print(capitales.values()) # values: Nos muestra los valores del diccionario
```

```python
print(capitales.items()) # items: Nos muestra el valor tanto las llaves del diccionario completo
```

```python
for key,value in capitales.items(): # es otra forma de mostrar todos los pares, para leer un diccionario tenemos que crear dos variables en lugar de una,
un valor para las llaves el for key,value (esas serian las dos variables) y luego poner como siempre el in (nombre de la variable)
  print(key,value)
```
