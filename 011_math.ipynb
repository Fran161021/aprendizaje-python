```python
import math # math: modulo para acceder a esras funciones
```

```python
PI = - 3.14
print(round(PI)) # round: redondea el número que le pasemos.
```

```python
print(math.ceil(PI)) # ceil: redondea un número hacía arriba
```

```python
print(math.floor(PI)) # floor: redondea un número hacía abajo.
```

```python
print(abs(PI)) # abs: nos da el valor absoluto, si le pasamos el valor de la variable negativo lo transforma en positivo.
```

```python
print(pow(PI,2)) # pow: es para elevar nuestro numero a la potencia que queramos.
```

```python
print(math.sqrt(420)) # sqtr: para sacar la raiz cuadrada acompañado de math
```
