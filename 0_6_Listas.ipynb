METODOS DE LISTAS

```python
lenguajes = ["Python", "Ruby", "PHP", "javascript", "Java"]
```

```python
lenguajes.insert(3, "Go")
```

```python
print(lenguajes)
```

```python
lenguajes.insert(0,"C")
print(lenguajes) # insert: inserta cualquier elemnto que pongamos
```

```python
lenguajes.remove("Ruby")
print(lenguajes) # remove: sirve para eliminar cualquier elemento
print("PHP" in lenguajes) # in: para saber si ese elemento está en la lista, si es así, true,sino false
```

```python
print(len(lenguajes)) # len: muestra cuantos elementos tiene nuestro listado
```

```python
lenguaje.clear() # clear, hace que se muestre vacío
```
