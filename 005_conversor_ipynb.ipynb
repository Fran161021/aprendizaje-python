EJERCICIO CONVERSOR DE TEMPERATURA

```python
temperatura = float(input("Ingresa la temperatura:")) # Se puede poner float ahí sin necesidad de crear otra variable, se puede poner tambien int,etc
escala = input("Es Fahrenheit(F) o es Celsius(C)?:").lower() # Tenemos que pasarlo todo a mayúsculas o minúsculas, en este cao mayúsculas
```

```python
if escala == "f":
   celsius = (temperatura - 32) * 5/9   # Debemos crear la variable celsius y esa es la fórmula para que se ponga en grado celsius
   print(celsius)
elif escala == "c":
  fahrenheit = temperatura * 1.8 + 32 #Esta es la formula para pasarlo a fahrenheit
  print(fahrenheit)
else:
  print("Escala incorrecta")
```
