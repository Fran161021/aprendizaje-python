```python
bebidas = ["café","cocacola","té"]
cena = ["Pizza", "Hamburguesas", "hot dog"]
postres = ["Pastel","Helado"]
comida = [bebidas,cena,postres]
print(comida[2][1])
```

He agrupado las 3 variables en una sola "comida" creando asi una lista
con el nombre de las 3 variables que hemos creado antes.
Si quiero ver un elemento de las listas que he creado pongo un indice
con [] y dentro se pone el numero de la lista que quiera ejecutar
luego si quiero ver un elemento dentro de esa lista
vuelvo a poner otro [] con el numero del elemento que quiera ver
