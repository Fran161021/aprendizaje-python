Un set es un conjunto sin orden ni indice que no permiten elementos duplicados pero si se puede comparar y unir otras operaciones con ellos y se usa para almacenar elementos de datos

```python
utensilios = {"tenedor","cucharas","cuchillo","cuchillo","cuchillo"}
utensilios.add("tijeras") # add: agrega un elemento pero de manera desordenada
utensilios.remove("cucharas") # remove: elelina el elemento que le digamos
utensilios.pop() # pop: elemina un elemento al azar
utensilios.clear() # deja vacío el conjunto
```

```python
utensilios = {"tenedor","cucharas","cuchillo","cuchillo","cuchillo"}
platos = {"plato","bol","taza"}
utensilios.update(platos) # une dos conjuntos
print(utensilios)
```

```python
utensilios = {"tenedor","cucharas","cuchillo","cuchillo","cuchillo"}
platos = {"plato","bol","taza","cucharas"}
print(utensilios.difference(platos)) # difference: descarta el elemento que sea similar al del otro conjunto en este caso cuchara
```

```python
utensilios = {"tenedor","cucharas","cuchillo","cuchillo","cuchillo"}
platos = {"plato","bol","taza","cucharas"}
print(utensilios.intersection(platos)) # intersection: nos devuelve el elemento repetido
```
