```python
mi_lista_original = [0,1,2,3,4,5,6,7]
print(mi_lista_original)

my_range = range(8)
print(list(my_range))

my_list = [i for i in range(8)]
print(my_list)

my_list = [i + 3 for i in range(8)]
print(my_list)

my_list = [i * 2 for i in range(8)]
print(my_list)

my_list = [i * i  for i in range(8)]
print(my_list)



```

```python
nombres = ["antonio","alberto","luis"]

nombres_grandes = [n.upper() for n in nombres]
print(nombres_grandes)
```

```python
correos_sucios = [" FRAN@GMAIL.COM ", " ANTONIO@YAHOO.ES", "PYTHON@OUTLOOK.COM "]
correos_limpios = [caja.strip().lower() for caja in correos_sucios] # strip: Borra y limpia los espacios en blancos invisibles
print(correos_limpios)

edades_clientes = [14,25,18,32,16,45,12,19]

permitidos = [p for p in edades_clientes if p >= 18] # el elemento de la izquierda siempre es obligatorio auqnue no se modifique nada y luego para el if tienes que poner
# como se llame el elemento que es el valor que va a coger de la iista de arriba
print(permitidos)
```

```python
edades_gente = [12,27,18,30,15,49,12,21]
edades_permitidas = [e for e in edades_gente if e >= 18]
print(edades_permitidas)

def calcular_iva(precio):
  return precio * 1.21
precios_sin_iva= [10,50,100,200]
facturas_finales = [calcular_iva(p) for p in precios_sin_iva]
print(facturas_finales)
```

```python

anos_nacimiento = [2004,2011,1998,2009,2005,2012,1995]
validos = [v for v in anos_nacimiento if v <= 2005]
print(validos)

precios_base = [10,25,50,100,12]
precios_totales = [p + 5 for p in precios_base] # P: Es el elemento luego p en el bucle va a coger el elemento de la lista de arriba y va a empezar a leer de izq a derecha
print(precios_totales)

def crear_referencia(REF):
  return f"REF-{REF}" # si tienes que ponerle un texto o algo en el return lo pones con f y llaves {}
numeros_id = [102,505,301,889]
codigos_finales = [crear_referencia(r) for r in numeros_id]
print(codigos_finales)
```

```python
litros_repostados = [15,62,8,45,70,12,55]
grandes_depositos = [l for l in litros_repostados if l >= 50]
print(grandes_depositos)

def generar_vuelo(IB):
  return f"IB-{IB}"
codigos_vuelos = [230,405,112,890]
panel_aeropuerto = [generar_vuelo(c) for c in  codigos_vuelos]
print(panel_aeropuerto)
```
