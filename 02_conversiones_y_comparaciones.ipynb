```python
n1 = 10
```

```python
n2 = 15
```

CONVERSION DE TIPOS

```python
resultado = input("Ingresa tu edad:") # pides al cliente que ingrese los datos con imput
numero = int(resultado) # luego lo domas con int para que sea numero entero
print(numero + 2) # imprimes con la variable que has domado a la otra
```

```python
edad = input("Cual es tu edad:")
print(type(edad))
conversion = int(edad)
print(conversion)
```
