INDEXACIÓN

```python
nombre = 'francisco javier guisado fernández'
if nombre[0].islower():
  nombre = nombre.capitalize()
  print(nombre)
```

```python
nombre = 'francisco javier guisado fernández'
primer_nombre = nombre[:4].upper()
print(primer_nombre)
```

```python
nombre = 'francisco javier guisado fernández'
apellido = nombre[5:].lower()
print(apellido)
```

```python
nombre = 'francisco javier guisado fernández'
ultimo_caracter = nombre[-1]
print(ultimo_caracter)
```
