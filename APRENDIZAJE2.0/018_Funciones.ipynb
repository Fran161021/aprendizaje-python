```python
def sum_two_numbers(number_one, number_two):
  print(number_one + number_two)
sum_two_numbers(4,2)
sum_two_numbers(66,39)
sum_two_numbers(1600,1700)

def suma_dos_numeros(primer_numero,segundo_numero):
  print("La suma de los dos números es de ", primer_numero + segundo_numero)
suma_dos_numeros(250000,60000)


def suma_numeros(un_numero,segundo_numero):
  print(un_numero + segundo_numero)
suma_numeros(16,16)
```

```python
def resta_tres_numeros(primer_numero,segundo_numero,tercer_numero):
  print(primer_numero - segundo_numero - tercer_numero)
resta_tres_numeros(15,2,4)

def otra_resta(primer_numero,segundo_numero):
  print(primer_numero - segundo_numero)
otra_resta(50,36)
```
