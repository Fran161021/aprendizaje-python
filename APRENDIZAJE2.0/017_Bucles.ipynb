```python
una_condicion = 10
while una_condicion < 20:
  una_condicion += 1
  if una_condicion == 15:
    print("Se detiene la ejecución")
    break
  print(una_condicion)
  # proceso: una condicion vale 10, 10 es menor que 20? si, 10 + 1 = 11, 11 es igual a 15 ? no, asi hasta llegar a 15 que es cuando se detendrá y se romperá el bucle
```

```python
mi_variable = 2 + 2
while mi_variable < 8:
  mi_variable += 2
  if mi_variable == 8:
    print("Detenemos la ejecución del bucle")
    break
  print(mi_variable)
```
