Función: Es un bloque de códigoque se ejecuta solo cuando se lo llama

```python
def saludo(): #def: es para definir la función
  print("Hola")
saludo() # una función solo funciona cuando se le llama
```

```python
def saludo():  # también podemos mandarle información y la función puede hacer algo con eso que recibe
  print("Hola")
  print("que tengas un buen dia")
saludo()
```

```python
def saludo(nombre): # En este paréntesis tenemo que declarar una variable temporal porque solamente se puede usar en el bloque de la función
  print("Hola " + nombre)
  print("que tengas un buen dia")
saludo("Fran") # Esto se llama argumento a la información que enviamos a una función
```

También podemos pasarle una variable

```python
def saludo(nombre2): #Esta variable junto a la del print es distinta a la de abajo
  print("Hola " + nombre2)
  print("que tengas un buen dia")
nombre = "Fran"
saludo(nombre) # Ahora si le pasamos la variable
```

```python
def saludo(nombre2):
  print("Hola " + nombre2)
  print("que tengas un buen dia")
nombre = input("Ingresa tu nombre: ") # También podemos pedirle que el usuario ingrese su nombre
saludo(nombre)
```

```python
def saludo(primer_nombre,apellido):
  print("Hola " + primer_nombre + " " + apellido)
  print("que tengas un buen dia")
nombre = input("Ingresa tu nombre: ") # También podemos pedirle que el usuario ingrese su nombre
saludo(nombre, "Javier") # podemos añadiroe más argumentos pero para que funcione también tenemos que añadir variables tempotales
```

```python
def saludo(primer_nombre,apellido, edad):
  print("Hola " + primer_nombre + " " + apellido)
  print("Tienes " + str(edad) + " años")
  print("que tengas un buen dia")
nombre = "Francisco"
saludo(nombre, "Javier", 23) # también podemos pasarle nuesra edad y para ello creamos otra variable temporal y lo pasamos a str
```
