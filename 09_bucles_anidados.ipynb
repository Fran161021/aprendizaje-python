Un bucle anidado es cuando tienes un blucle dentro de otro bucle

```python
filas = int(input("Cuantas filas quieres? "))
columnas = int(input("Cuantas columnas quieres? "))
simbolo = input("Ingresa un simbolo ")
for i in range(filas):
  for j in range(columnas): # el primer for es con "i", el segundo tiene que ser con otra nomenclatura para que funcione
    print(simbolo, end="")
  print() # Este print tiene que estar alineado con el segundo "for" para que funcione bien
```

BREAK CONTINUE PASS

```python
nombre = ""
```

```python
while True:
  nombre = input("Ingresa tu nombre: ")
  if nombre != "":
    break # break: Rompe y sale del bucle while
```

```python
telefono = "123-456-789"
```

```python
for i in telefono:
  if i == "-":
    continue # continue: sirve para decirle que si ve un guión lo ignore si en vez de "-" fuese "4" ignoraria el numero 4 solamente
  print(i, end="")
```

```python
for i in range(1,21):
  if i == 13:
    pass # pass: significa pasa de largo el número 13
  else:
    print(i)
```

Break se suele ver mucho con while true porque es la unica forma de romper un bucle infinito.
Continue: No apaga el bucle, se ve muco en "for", para revisar listas o texto elemento a elemento y continue es perfecto para filtrar lo que no queremos
Pass: es un relleno para que Python no de error si dejas un bloque vacío.
