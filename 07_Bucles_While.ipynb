While para acceder a los elementos de un listado

```python
i = 0 # los listados comienzan con los indices 0
while i < len(lenguajes):
  print(lenguajes[i])
  i = i + 1
```

BUCLES FOR

```python
lenguajes = ["Python","Ruby","PHP","JavaScript","Java"]
```

```python
for lenguaje in lenguajes: # es mas rapido que while a la hora de iterar elementos dentro de un listado
  print(lenguaje)
```
