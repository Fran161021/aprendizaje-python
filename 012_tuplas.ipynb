Las tuplas son una lista de elementos que no se pueden modificar, se puede usar para crear un registro de estudiantes

```python
estudiantes = ('Alex',22,'H','Anto',22,'M') # Es como en las listas pero en vez de corchetes, paréntesis
print(estudiantes.count(22)) # count: cuenta cuantas veces sale una palabara o lo que le pongamos
```

```python
estudiantes = ('Alex',22,'H','Anto',22,'M')
print(estudiantes.index('H')) # index: nos dice la posición en la que se encuentra algo específico en este caso Hcomienza contando desde el 0
```

```python
for x in estudiantes:
  print(x) # nos muestra el valor almacenado dentro de esa variable
```

```python
if 'Alex' in estudiantes: # se puede usar para comprobar un valor específico está en la tupla
  print("Alex está aquí")
```
