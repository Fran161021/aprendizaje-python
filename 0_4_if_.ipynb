```python
puntuacion = 97
```

```python
if puntuacion >= 95:
  print("Aprobado con honores")
elif puntuacion >= 50:
  print("Alumno aprobado")
else:
  print("Suspenso")
print("Fuera del if") # Se va a ejecutar siempre ya que está fuera del sangrado
```

Se ejecuta de arriba a abajo, si el if es false, pasará al elif, si llega a ser false también, pasará al else. Y si el if es true los otros no se ejecutarán
